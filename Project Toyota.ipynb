Load Dataset

```python
from google.colab import drive
import pandas as pd

# Load dataset
df = pd.read_csv('/content/drive/MyDrive/SI DATA SCIENTIST CAKAP/toyota.csv')
```

Analytical DataFrame

```python
#Lihat 5 baris pertama
print(df.head())
```

```python
#Lihat informasi DataFrame
print(df.info())
```

Check Duplicate & Handle Duplicate

```python

# Cek duplikat
duplicates = df.duplicated().sum()
print(f'Jumlah duplikat: {duplicates}')

# Hapus duplikat jika ada
df = df.drop_duplicates()
```

Filter DataFrame

```python
#Memfilter berdasarkan kolom 'Price' lebih dari 15000
filtered_df = df[df['price'] > 15000]
print(filtered_df.head())
```

Group DataFrame

```python
#mengelompokkan berdasarkan kolom 'Model' dan menghitung rata-rata 'Price'
grouped_df = df.groupby('year')['price'].mean().reset_index()
print(grouped_df)
```

Replace Dataframe

```python
#DataFrame awal
#Nilai awal kolom 'fuelType'
data = {
    'model': ['GT86', 'GT86', 'GT86', 'GT86'],
    'year': ['2016', '2017', '2015', '2017'],
    'fuelType': ['Petrol', 'Petrol', 'Diesel', 'Petrol'],
    'price': [16000, 15995, 13998, 18998]
}

df = pd.DataFrame(data)
print("DataFrame Awal:")
print(df)

#Mengganti nilai dalam kolom 'fuelType'
df['fuelType'] = df['fuelType'].replace({
    'Petrol': 'Gasoline',
    'Diesel': 'petrol'
})

#Mencetak DataFrame setelah penggantian
print("\nDataFrame Setelah Penggantian:")
print(df)
```
